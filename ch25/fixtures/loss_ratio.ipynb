# Basement water claims: loss ratio by state
Quarterly review for the pricing committee.

```
claims = read_table('claims_basement_water')
claims = claims[claims.peril.isin(['sewer_backup','seepage','surface_water'])]
```

```
premiums = read_table('earned_premium_ho3')
```

```
claims['wbu'] = claims.endorsements.str.contains('HO 04 95')
```

```
joined = claims.merge(premiums, on=['state','year'])
```

```
by_state = joined.groupby(['state','year']).agg(count=('claim_id','size'), premium=('earned','sum'), losses=('paid','sum'))
by_state['loss_ratio'] = by_state.losses / by_state.premium
by_state
```

## Endorsed vs not endorsed

```
split = joined.groupby(['state','wbu']).paid.sum().unstack()
split
```
